```python
from collections import Counter
from scipy.stats import entropy

def shannon_entropy(s):
    """Calculate the Shannon entropy of a string."""
    if not s:
        return 0
    counts = Counter(s)
    entropy_value = entropy(list(counts.values()), base=2)
    return entropy_value

# Feature Engineering
df['n_hyphens_in_domain'] = df['Domain'].apply(lambda x: str(x).count('-'))
df['hyphen_to_domain_length_ratio'] = df['n_hyphens_in_domain'] / df['DomainLength']
df['domainToURLLengthRatio'] = df['DomainLength'] / df['URLLength']
df['domain_entropy'] = df['Domain'].apply(lambda x: shannon_entropy(str(x)))
df['is_international_domain'] = df['Domain'].apply(lambda x: 1 if str(x).startswith('xn--') else 0)

df
```
